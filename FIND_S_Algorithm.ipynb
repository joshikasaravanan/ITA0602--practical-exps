```python
import pandas as pd

# Read the uploaded Kaggle dataset
df = pd.read_csv('/content/enjoysport.csv')

print("Training Dataset:")
print(df)

# FIND-S Algorithm
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

hypothesis = None

for i in range(len(df)):
    if str(y.iloc[i]).strip().lower() == 'yes':
        example = X.iloc[i].tolist()

        if hypothesis is None:
            hypothesis = example.copy()
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != example[j]:
                    hypothesis[j] = '?'

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```
