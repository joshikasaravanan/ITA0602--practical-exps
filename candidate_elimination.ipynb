```python
import numpy as np
import pandas as pd
data=pd.read_csv('/kaggle/input/candidateelimination/findS.csv')
data=pd.DataFrame(data)

print(data)
```

```python
concepts = np.array(data.iloc[:,0:-1])

target = np.array(data.iloc[:,-1])

print(target)

print(concepts)
```

```python
def learn(concepts, target):

    specific_h = concepts[0].copy()

    print("Initialization of specific_h and general_h")

    print("specific_h: ",specific_h)

    general_h = [["?" for i in range(len(specific_h))] for i in range(len(specific_h))]

    print("general_h: ",general_h)

    print("concepts: ",concepts)

    for i, h in enumerate(concepts):

        if target[i] == "yes":

            for x in range(len(specific_h)):

                #print("h[x]",h[x])

                if h[x] != specific_h[x]:

                    specific_h[x] = '?'

                    general_h[x][x] = '?'

        if target[i] == "no":

            for x in range(len(specific_h)):

                if h[x] != specific_h[x]:

                    general_h[x][x] = specific_h[x]

                else:

                    general_h[x][x] = '?'

    print("\nSteps of Candidate Elimination Algorithm: ",i+1)

    print("Specific_h: ",i+1)

    print(specific_h,"\n")

    print("general_h :", i+1)

    print(general_h)

    indices = [i for i, val in enumerate(general_h) if val == ['?', '?', '?', '?', '?', '?']]

    print("\nIndices",indices)

    for i in indices:

        general_h.remove(['?', '?', '?', '?', '?', '?'])

    return specific_h, general_h

s_final,g_final = learn(concepts, target)

print("\nFinal Specific_h:", s_final, sep="\n")

print("Final General_h:", g_final, sep="\n")
```

```python

def candidate_elimination(concepts, target):
    n = len(concepts[0])

    S = concepts[0].copy()
    G = [['?' for _ in range(n)]]

    for i in range(len(target)):
        x = concepts[i]

        if target[i] == 'Yes':
            for j in range(n):
                if S[j] != x[j]:
                    S[j] = '?'

            G = [g for g in G if all(
                g[j] == '?' or g[j] == x[j]
                for j in range(n)
            )]

        else:
            new_G = []
            for g in G:
                if all(g[j] == '?' or g[j] == x[j]
                       for j in range(n)):
                    for j in range(n):
                        if S[j] != '?' and S[j] != x[j]:
                            h = g.copy()
                            h[j] = S[j]
                            new_G.append(h)
                else:
                    new_G.append(g)
            G = new_G

    print("Final Specific Boundary S:", S)
    print("Final General Boundary G:")
    for g in G:
        print(g)

candidate_elimination(concepts, target)
```
